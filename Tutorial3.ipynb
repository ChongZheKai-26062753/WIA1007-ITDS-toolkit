```python
# ==============================================================================
# Tutorial 03: Tidy Data, Data Characteristics
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive

# ------------------------------------------------------------------------------
# Q1. Find some real dataset. Load it on Colab, Kaggle etc. Try to visualize some of it
# ------------------------------------------------------------------------------

# Mount Google Drive
drive.mount('/content/drive')

# Load dataset from Drive
# Update file name if yours is slightly different (e.g. 'zomato.csv')
file_path = '/content/drive/MyDrive/archive/zomato.csv'
df = pd.read_csv(file_path, encoding='latin-1')  # latin-1 encoding handles special characters in restaurant names

# Quick preview
print("--- FIRST 5 ROWS ---")
display(df.head())

# Visualizations
plt.figure(figsize=(12, 5))

# 1. Rating Distribution Plot
plt.subplot(1, 2, 1)
sns.histplot(df['Aggregate rating'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Restaurant Aggregate Ratings')
plt.xlabel('Aggregate Rating')
plt.ylabel('Count')

# 2. Top Cities Bar Plot
plt.subplot(1, 2, 2)
top_cities = df['City'].value_counts().head(5)
sns.barplot(x=top_cities.values, y=top_cities.index, palette='Blues_r')
plt.title('Top 5 Cities by Number of Restaurants')
plt.xlabel('Number of Restaurants')

plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# Q2 & Q4. Evaluate if the data is good or not. Does it have enough samples?
# How many samples/observation does it have? Is it enough?
# ------------------------------------------------------------------------------
num_samples = df.shape[0]

print("\n--- SAMPLE EVALUATION ---")
print(f"Total Samples / Observations (Rows): {num_samples}")

if num_samples >= 1000:
    print("Evaluation: The dataset has plenty of samples for standard exploratory data analysis (EDA), visualization, and training predictive regression/classification models.")
else:
    print("Evaluation: Small sample size; might be prone to overfitting or sampling bias.")

# ------------------------------------------------------------------------------
# Q3. How many features does it have? Is it enough?
# ------------------------------------------------------------------------------
num_features = df.shape[1]

print("\n--- FEATURE EVALUATION ---")
print(f"Total Features (Columns): {num_features}")
print("Feature Names:", list(df.columns))
print("Evaluation: 21 features provide a rich mix of numerical ratings, price levels, geographical identifiers, and categorical attributes (e.g., Cuisines, Has Table booking, Has Online delivery). This is more than enough to analyze market trends and build models.")

# ------------------------------------------------------------------------------
# Q5. What are the datatypes does the dataset consist of?
# ------------------------------------------------------------------------------
print("\n--- DATA TYPES ---")
print(df.dtypes)

# Summary of data types breakdown
print("\nData Type Summary:")
print(df.dtypes.value_counts())

# ------------------------------------------------------------------------------
# Q6 & Q8. How many columns have null values? Which column, which rows?
# Does it have missing values? If yes, which column, which row.
# ------------------------------------------------------------------------------
print("\n--- MISSING / NULL VALUES ANALYSIS ---")

# Columns with null values
null_counts = df.isnull().sum()
cols_with_nulls = null_counts[null_counts > 0]

if not cols_with_nulls.empty:
    print(f"Number of columns with null values: {len(cols_with_nulls)}\n")
    print("Columns with null values and counts:")
    print(cols_with_nulls)

    # Locate specific row indices containing null values
    null_rows_indices = df[df.isnull().any(axis=1)].index.tolist()
    print(f"\nTotal rows containing at least one null value: {len(null_rows_indices)}")
    print(f"Sample row indices with missing values: {null_rows_indices[:10]}...")

    # Display the actual rows with missing values
    print("\nDisplaying actual missing value rows:")
    display(df[df.isnull().any(axis=1)].head())
else:
    print("No missing / null values found across the dataset.")

# ------------------------------------------------------------------------------
# Q7. Which column is the feature, which is the target output?
# ------------------------------------------------------------------------------
print("\n--- FEATURE & TARGET DEFINITION ---")

# Target Output (What we want to predict or analyze)
target = "Aggregate rating"  # Or "Average Cost for two" depending on task

# Predictor Features (Input variables)
features = [col for col in df.columns if col != target]

print(f"Target Variable (Output): '{target}'")
print("\nPredictor Features (Inputs):")
for f in features:
    print(f" - {f}")

# ------------------------------------------------------------------------------
# Q9. What kind of question can be asked, should we ask?
# Does the dataset suit the question we are asking?
# ------------------------------------------------------------------------------
print("\n--- RESEARCH QUESTIONS & DATASET SUITABILITY ---")
print("""
1. Research Questions to Ask:
   a) Does offering online delivery or table booking significantly increase a restaurant's aggregate rating?
   b) Which cuisines command the highest average cost for two across different cities?
   c) Can we accurately predict a restaurant's aggregate rating using its price range, city, votes, and online delivery status?

2. Dataset Suitability:
   - Yes, the dataset is highly suitable for these questions. It contains explicit boolean features ('Has Online delivery', 'Has Table booking'), continuous numeric targets ('Aggregate rating', 'Votes', 'Average Cost for two'), and categorical location identifiers ('City', 'Cuisines').
""")
```
