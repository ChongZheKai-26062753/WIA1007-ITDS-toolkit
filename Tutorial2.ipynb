```python
import numpy as np

# Create a 2x2 numpy array
array_2x2 = np.array([[1, 2], [3, 4]])
print(array_2x2)
```

```python
# Create a random 3x4 numpy array with integers from 1 to 20
random_int_array = np.random.randint(1, 21, size=(3, 4))
print(random_int_array)
```

```python
# Create a numpy array of ones (size 2x3)
ones_array = np.ones((2, 3))
print("Ones array:")
print(ones_array)

# Create a numpy array of zeros (size 3x3)
zeros_array = np.zeros((3, 3))
print("\nZeros array:")
print(zeros_array)
```

```python
# Create a random integer numpy array of size 6x3
original_array = np.random.randint(1, 20, size=(6, 3))
print("Original Array (6x3):")
print(original_array)

# Transpose the array to 3x6
transposed_array = original_array.T
print("\nTransposed Array (3x6):")
print(transposed_array)
```

```python
# Define dimensions
n, p, m = 4, 3, 2

# Create two random numpy arrays
array_A = np.random.randint(1,10,size=(n, p))
array_B = np.random.randint(1,10,size=(p, m))

# Compute the dot product
dot_product = np.dot(array_A, array_B)

print(f"Array A ({n}x{p}):")
print(array_A)
print(f"\nArray B ({p}x{m}):")
print(array_B)
print(f"\nDot Product ({n}x{m}):")
print(dot_product)
```
